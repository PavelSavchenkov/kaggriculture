## 1. Complete programmes first, bounded adaptation second
<div class="kg-prose"><div class="lead">Long-horizon farm work lives inside four complete action tapes. The router does not replan every turn. It changes the active tape only at two fixed dawn boundaries, using fields that are already observable at that moment. This keeps movement, planting, livestock care, logistics, and planned trading temporally coherent.</div></div>

## 2. Two public-state route decisions
<div class="kg-prose"><div class="lead">The season opens on route 0. At step 144, the number of already-unlocked YARN_STORE shops decides whether to continue route 0 or switch to route 1. At step 648, current public EGG market inventory selects the late route 2 or route 3. After the selected tape emits its action, repeated same-product SELLs are compacted into the first matching slot. At one frozen SELL-only timing point, step 550, an empty market gap is removed so the planned WOOL order is processed one slot earlier.</div></div>

```python
import pandas as pd

route_rules = pd.DataFrame([
    ("Opening", 0, "route 0", "start with the default complete programme"),
    ("Early branch", 144, "route 0 / 1", "YARN_STORE count > 0 → route 1; otherwise route 0"),
    ("Late branch", 648, "route 2 / 3", "public EGG market inventory ≤ 9,888 → route 2; otherwise route 3"),
    ("SELL compaction", "every step", "same route", "merge repeated SELLs of one product into its earliest live slot"),
    ("Timing cleanup", 550, "same route", "remove one no-op gap on this SELL-only turn; keep product and quantity unchanged"),
], columns=["decision", "step", "selected programme", "public rule"])
print(route_rules.to_string(index=False))
```

```python
from collections import Counter
from pathlib import Path
import contextlib, io, math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import HTML, Image, display

with contextlib.redirect_stdout(io.StringIO()), contextlib.redirect_stderr(io.StringIO()):
    from kaggle_environments import make

viz_env = make("kaggriculture", configuration={"episodeSteps": 720, "seed": 910019}, debug=False)
_agent_path = str(Path("main.py").resolve())
with contextlib.redirect_stdout(io.StringIO()), contextlib.redirect_stderr(io.StringIO()):
    viz_env.run([_agent_path, _agent_path])
assert len(viz_env.steps) == 720
assert int(viz_env.steps[-1][0].observation.step) == 719
assert [str(s.status) for s in viz_env.steps[-1]] == ["DONE", "DONE"]

FOREST = "#173322"; MUTED = "#64776b"; GRID = "#d7e5dc"
PALETTE = ["#2a8a54", "#168a9a", "#b78312", "#8b5fb2", "#c75b4f"]

def _show_table(frame, title, formats=None):
    f = frame.copy()
    if formats:
        for col, func in formats.items():
            if col in f.columns:
                f[col] = f[col].map(func)
    h = f.to_html(index=False, escape=True).replace('class="dataframe"','class="forest-table"')
    display(HTML(
        f'<div style="background:#fff!important;border:1px solid #d5e3da;border-radius:17px;padding:14px 16px;overflow-x:auto;box-shadow:0 5px 16px rgba(24,68,44,.05);color:#173322!important;color-scheme:light">'
        f'<div style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:800;color:#176b42;margin-bottom:9px">{title}</div>'
        '<style>.forest-table{border-collapse:collapse;width:100%;font-size:12.8px;background:#fff!important;color:#203a2c!important;color-scheme:light}.forest-table thead,.forest-table th{background:#edf5f0!important;color:#173322!important}.forest-table th{padding:9px 10px;text-align:left;border-bottom:1px solid #d6e3da!important}.forest-table tbody,.forest-table tr,.forest-table td{background:#fff!important;color:#203a2c!important}.forest-table tbody tr:nth-child(even) td{background:#f7faf8!important}.forest-table tbody tr:hover td{background:#edf7f1!important}.forest-table td{padding:9px 10px;border-bottom:1px solid #e5eee8!important}.forest-table tr:last-child td{border-bottom:0!important}</style>'
        + h + '</div>'
    ))

def _show_kpis(items, eyebrow="Executed season"):
    cards=[]
    for label,value,note,accent in items:
        cards.append(f'<div style="background:#fff;border:1px solid #d5e3da;border-top:4px solid {accent};border-radius:16px;padding:13px 14px;color:#173322;box-shadow:0 4px 14px rgba(24,68,44,.04)"><div style="font-size:10px;letter-spacing:.11em;text-transform:uppercase;font-weight:800;color:#64786c">{label}</div><div style="font-size:22px;font-weight:850;margin:3px 0 2px">{value}</div><div style="font-size:12px;color:#687970">{note}</div></div>')
    display(HTML(f'<div style="font-size:11px;letter-spacing:.11em;text-transform:uppercase;font-weight:800;color:#176b42;margin:16px 0 8px">{eyebrow}</div><div style="display:grid;grid-template-columns:repeat(auto-fit,minmax(150px,1fr));gap:9px;margin-bottom:12px">'+''.join(cards)+'</div>'))

def _forest_axes(ax, title, xlabel="", ylabel=""):
    ax.set_title(title, loc="left", color=FOREST, fontsize=12.5, fontweight="bold", pad=10)
    ax.set_xlabel(xlabel, color=MUTED, fontsize=9); ax.set_ylabel(ylabel, color=MUTED, fontsize=9)
    ax.tick_params(colors=MUTED, labelsize=8.2)
    ax.spines["top"].set_visible(False); ax.spines["right"].set_visible(False)
    ax.spines["left"].set_color(GRID); ax.spines["bottom"].set_color(GRID)
    ax.set_facecolor("#fbfdfb")

def _show_fig(fig, quality=70):
    buf=io.BytesIO(); fig.savefig(buf, format="jpeg", dpi=92, bbox_inches="tight", facecolor="white", pil_kwargs={"quality": quality})
    plt.close(fig); display(Image(data=buf.getvalue()))

def _plant_count(farm):
    n=0
    for row in farm.get("tiles", []):
        for tile in row:
            if isinstance(tile, dict) and tile.get("kind") == "PLANT": n += 1
    return n

def _pasture_count(farm):
    n=0
    for row in farm.get("tiles", []):
        for tile in row:
            if isinstance(tile, dict) and tile.get("kind") == "PASTURE": n += 1
    return n

def _tile_cat(tile):
    if tile == "LOCKED": return "LOCKED"
    if tile is None: return "EMPTY"
    if isinstance(tile, dict):
        if tile.get("kind") == "PLANT": return tile.get("crop", "PLANT")
        return tile.get("kind", "OTHER")
    return "OTHER"

def show_motion(env, seat=0, stride=12, duration=18):
    frame_ids=list(range(0,len(env.steps),stride))
    if frame_ids[-1] != len(env.steps)-1: frame_ids.append(len(env.steps)-1)
    obs=[env.steps[i][seat].observation for i in frame_ids]
    pid=int(obs[0].player); farms=[o.farms[pid] for o in obs]
    rows=len(farms[0]["tiles"]); cols=len(farms[0]["tiles"][0]); cell=24; gap=2; x0=32; y0=67
    colors={"LOCKED":"#263b31","EMPTY":"#e9f0eb","WHEAT":"#e5b94e","CARROT":"#e99a4a","TOMATO":"#d96b5f","STRAWBERRY":"#d85c70","MELON":"#61b873","PASTURE":"#4b9da5","WEED":"#a66bb5","COOP":"#9b795e","OTHER":"#a9b8af"}
    p=['<div style="background:#fff;border:1px solid #d5e3da;border-radius:22px;padding:12px;margin:14px 0;box-shadow:0 8px 24px rgba(24,68,44,.06);overflow-x:auto;color:#173322"><svg viewBox="0 0 930 365" width="100%" style="min-width:730px;max-width:1000px;display:block;margin:auto" xmlns="http://www.w3.org/2000/svg">']
    p+=['<rect x="8" y="8" width="914" height="349" rx="21" fill="#f8fbf9" stroke="#d6e2da"/><text x="30" y="34" font-size="17" font-weight="800" fill="#173322">Executed farm in motion</text><text x="30" y="52" font-size="11" fill="#607168">Compressed from the exact 720-turn self-play trajectory · four-route replay · compact SELL packing · step-550 timing</text>']
    # progress rail
    p.append('<line x1="530" y1="36" x2="870" y2="36" stroke="#d1ddd5" stroke-width="7" stroke-linecap="round"/>')
    p.append(f'<circle cx="530" cy="36" r="7" fill="#fff" stroke="#173322" stroke-width="2"><animate attributeName="cx" values="530;870" dur="{duration}s" repeatCount="indefinite"/></circle>')
    p.append('<text x="530" y="56" font-size="9" fill="#607168">start</text><text x="852" y="56" font-size="9" fill="#607168">close</text>')
    # tiles
    for r in range(rows):
        for c in range(cols):
            vals=[colors.get(_tile_cat(f["tiles"][r][c]),colors["OTHER"]) for f in farms]
            x=x0+c*cell; y=y0+r*cell
            p.append(f'<rect x="{x}" y="{y}" width="{cell-gap}" height="{cell-gap}" rx="4" fill="{vals[0]}"><animate attributeName="fill" values="{";".join(vals)}" dur="{duration}s" calcMode="discrete" repeatCount="indefinite"/></rect>')
    # farmer
    xs=[]; ys=[]
    for f in farms:
        rr,cc=f["farmer"]; xs.append(x0+cc*cell+(cell-gap)/2); ys.append(y0+rr*cell+(cell-gap)/2)
    p.append(f'<circle cx="{xs[0]:.1f}" cy="{ys[0]:.1f}" r="7.5" fill="#fff" stroke="#173322" stroke-width="2.5"><animate attributeName="cx" values="{";".join(f"{v:.1f}" for v in xs)}" dur="{duration}s" repeatCount="indefinite"/><animate attributeName="cy" values="{";".join(f"{v:.1f}" for v in ys)}" dur="{duration}s" repeatCount="indefinite"/></circle>')
    # right readout
    p.append('<text x="310" y="86" font-size="10" font-weight="800" letter-spacing="1.2" fill="#607168">HOW TO READ THIS</text>')
    notes=[("Fields appear","productive tiles replace empty ground"),("Pastures persist","livestock infrastructure becomes long-lived"),("Farmer circulates","the white marker follows the selected route tape"),("Final market adapts","same-product SELLs are compacted; one step-550 market gap is removed")]
    yy=108
    for a,b in notes:
        p.append(f'<circle cx="314" cy="{yy-3}" r="4" fill="#2a8a54"/><text x="327" y="{yy}" font-size="12" font-weight="800" fill="#173322">{a}</text><text x="327" y="{yy+17}" font-size="10.5" fill="#607168">{b}</text>'); yy+=48
    p.append('<text x="310" y="310" font-size="9.5" fill="#607168">Tile colors are descriptive only; the animation does not drive the agent.</text>')
    p.append('</svg></div>'); display(HTML(''.join(p)))

# Build daily summaries from seat 0 only.
days=np.arange(1,31)
daily_farm_actions=[]; daily_market_orders=[]; daily_bank=[]; daily_hands=[]; daily_plants=[]; daily_pastures=[]
products=["WHEAT","STRAWBERRY","MILK","FERTILIZER"]
daily_prices={p:[] for p in products}; mix=Counter()
movement={"NORTH","SOUTH","EAST","WEST"}; crop={"PLANT","WATER","HARVEST","FERTILIZE","DIG"}; livestock={"FEED","CARE","COLLECT_FERTILIZER"}
for d in range(30):
    a0=d*24; a1=min((d+1)*24,len(viz_env.steps))
    fops=0; mops=0
    for step in range(a0,a1):
        action=viz_env.steps[step][0].action or {}
        units=[action.get("farmer")] + list(action.get("hands",[]) or [])
        for item in units:
            if not item: continue
            op=item[0]
            if op=="PASS": mix["Pass"]+=1
            else:
                fops+=1
                if op in movement: mix["Movement"]+=1
                elif op in crop: mix["Crop work"]+=1
                elif op in livestock: mix["Livestock"]+=1
                elif op in {"PICKUP","DROP","PLACE"}: mix["Logistics"]+=1
                elif op.startswith("BUILD"): mix["Building"]+=1
                else: mix["Other farm"]+=1
        orders=list(action.get("market",[]) or []); mops+=len(orders)
        for order in orders:
            if not order: continue
            if order[0]=="SELL": mix["Market sales"]+=1
            else: mix["Investment / buying"]+=1
    daily_farm_actions.append(fops); daily_market_orders.append(mops)
    end=viz_env.steps[a1-1][0].observation; farm=end.farms[0]
    daily_bank.append(float(farm["money"])); daily_hands.append(len(farm.get("hands",[]))); daily_plants.append(_plant_count(farm)); daily_pastures.append(_pasture_count(farm))
    for pdt in products: daily_prices[pdt].append(float(end.market["prices"][pdt]))

show_motion(viz_env)
verification=pd.DataFrame([{"turns":len(viz_env.steps),"final step":int(viz_env.steps[-1][0].observation.step),"seat 0":str(viz_env.steps[-1][0].status),"seat 1":str(viz_env.steps[-1][1].status),"main.py SHA":"131504becf57a8ad…"}])
_show_table(verification,"Exact-agent execution check")
```
